## Which is the most popular Restaurant category?

```python
df_copy.head(2)
```

```python
# Grouping the continuous 'Votes' column into logical ranges for a clean pie chart representation
vote_groups = pd.cut(
    df_copy['Votes'],
    bins=[-1, 0, 100, 500, float('inf')],
    labels=['No Votes (0)', 'Low Votes (1-100)', 'Medium Votes (101-500)', 'High Votes (>500)']
)

# Calculate percentages and plot
plt.figure(figsize=(8, 8))
vote_groups.value_counts().plot.pie(
    autopct='%1.1f%%',
    startangle=140,
    colors=['#ff9999','#66b3ff','#99ff99','#ffcc99'],
    explode=(0.05, 0.05, 0.05, 0.05)
)
plt.title('Distribution of Restaurant Vote Categories', fontsize=14, fontweight='bold')
plt.ylabel('')  # Remove default y-axis label for cleaner look
plt.show()
```

## Which Category has largest number of installations??

```python
df_cat_installs = df_copy.groupby(['Cuisines'])['Votes'].sum().sort_values(ascending=False).reset_index()
df_cat_installs['Votes_Millions'] = df_cat_installs['Votes']/1000000 # converting into millions
df2 = df_cat_installs.head(10)
plt.figure(figsize=(14,10))
sns.set_context("talk")
sns.set_style("darkgrid")

ax = sns.barplot(x='Votes_Millions', y='Cuisines', data=df2)
ax.set_xlabel('No. of Votes in Millions')
ax.set_ylabel('')
ax.set_title("Most Popular Cuisine Categories by Votes", size=20)
```

## What are the Top 5 most visited Restaurant in Each popular Categories ??

```python
# We select the top 4 most popular cuisines based on total votes
popular_cuisines = ['North Indian', 'North Indian, Mughlai', 'North Indian, Chinese', 'Cafe']

# Filter the copy of our dataset
dfa = df_copy[df_copy['Cuisines'].isin(popular_cuisines)]
sns.set_context("talk")
sns.set_style("darkgrid")

plt.figure(figsize=(20, 16))

for i, cuisine in enumerate(popular_cuisines):
    # Get top 5 restaurants for each cuisine category based on votes
    df_cuisine = dfa[dfa['Cuisines'] == cuisine].sort_values('Votes', ascending=False).head(5)

    plt.subplot(2, 2, i+1)
    sns.barplot(data=df_cuisine, x='Votes', y='Restaurant Name', palette='viridis')
    plt.xlabel('Number of Votes')
    plt.ylabel('')
    plt.title(f'Top 5 in {cuisine}', size=16)

plt.tight_layout()
plt.show()
```

##Which Restaurant has 5 star rating

```python
rating = df_copy.groupby(['Restaurant Name', 'City', 'Aggregate rating'])['Votes'].sum().sort_values(ascending=False).reset_index()

toprating_apps = rating[rating['Aggregate rating'] == 4.9]
print("Number of Excellent rated Restaurants:", toprating_apps.shape[0])
toprating_apps.head(1)
```
